1) Задача 8: Создайте базовый класс Vehicle с полями марка и модель. Реализуйте подклассы Car и Bicycle, которые добавляют свои специфические методы, например, метод honk() для автомобиля.


```python
class Vehicle:
  def __init__(self,marka,model):
    self.marka = marka
    self.model = model
  def display_info(self):
    print(f"Транспортик марки {self.marka}  и  модельки {self.model}")

class Car(Vehicle):
  def signal(self):
    print("у машинки сигнализация")

class Bicycle(Vehicle):
  def zvonok(self):
    print("у велосипеда звонок")

car = Car("Toyota", "Camry")
bicycle = Bicycle("Stels", "Navigator")
car.display_info()
car.signal()
bicycle.display_info()
bicycle.zvonok()
```

2)Задача 8. Определите класс Car, который наследуется от класса Vehicle. Класс Car должен добавлять методы для работы с автомобильными функциями.

```python
class Vehicle:
  def __init__(self,marka,model):
    self.marka = marka
    self.model = model
  def display_info(self):
    print(f"Транспортик марки {self.marka}  и  модельки {self.model}")

class Car(Vehicle):
  def signal(self):
    print("у машинки сигнализация")
  def dvigatel(self):
    print("двигатель рычит")
  def svet(self):
    print("фары светят")


car = Car("Toyota", "Camry")

car.display_info()
car.signal()
car.dvigatel()
car.svet()
```

3) Задача 8. Создать класс Car, у которого поля brand, model и year будут приватными. Реализовать методы для доступа к этим полям и изменения модели с проверкой на валидность введенных данных.


```python
class Car:
    def __init__(self, marka, model, god):
        if marka.strip() != "" and model.strip() != "" and god >= 1900:
            self.__marka = marka
            self.__model = model
            self.__god = god
        else:
            print("неправильные данные")

    def get_marka(self):
        return self.__marka
    def get_model(self):
        return self.__model
    def get_god(self):
        return self.__god

    def set_model(self, new_model):
        if new_model.strip() != "":
            self.__model = new_model
        else:
            print("нет такой модельки")

car = Car("Toyota", "Camry", 2004)
print("Марка:", car.get_marka())
print("Модель:", car.get_model())
print("Год:", car.get_god())

car.set_model("Corolla")
print("Новая модель:", car.get_model())
```
