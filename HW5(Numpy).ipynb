Задача 1 (создание массива, dtype, shape/size/ndim). В аэропорту ведётся учёт стоимости билетов на 7 рейсов: [10000, 35000, 70000, 15000, 25000, 50000, 12000] рублей. Создайте массив NumPy из этих значений. Выведите его dtype, shape, size, ndim. Дополнительно создайте второй массив из тех же значений, но с типом float32, и сравните dtype и nbytes с первым. Ожидаемый вывод: для первого массива dtype=int64, shape=(7,), size=7, ndim=1; для второго — dtype=float32, nbytes=28. В комментариях поясните: почему NumPy выбрал именно int64 для первого массива; чем shape отличается от size; почему shape у одномерного массива — это кортеж из одного элемента (7,), а не (7); что произойдёт при обращении prices[7] и почему; как nbytes связан с itemsize и size.

```python
import numpy as np

prices = np.array([10000, 35000, 70000, 15000, 25000, 50000, 12000])

print("Первый массив:", prices)
print("dtype:", prices.dtype)
print("shape:", prices.shape)
print("size:", prices.size)
print("ndim:", prices.ndim)
print("nbytes:", prices.nbytes)

prices_float = np.array([10000, 35000, 70000, 15000, 25000, 50000, 12000], dtype=np.float32)

print("Второй массив:", prices_float)
print("dtype:", prices_float.dtype)
print("nbytes:", prices_float.nbytes)

# 1. NumPy выбрал int64, потому что все числа целые,
#    а на 64-битной платформе это стандартный целочисленный тип.
# 2. shape показывает форму массива, а size — общее количество элементов.
# 3. (7,) — кортеж из одного элемента. (7) — просто число 7 в скобках.
# 4. prices[7] вызовет IndexError, так как индексы идут от 0 до 6.
# 5. nbytes = itemsize * size.
#    Для int64: 8 * 7 = 56 байт.
#    Для float32: 4 * 7 = 28 байт.
```

Задача 8 (случайные числа, воспроизводимость). Аэропорт моделирует данные для симуляции пассажиропотока. Создайте генератор rng = np.random.default_rng(seed=42) и сгенерируйте: 10 равномерных значений rng.random(10) — вероятности покупки билета; 10 нормальных значений rng.normal(15000, 5000, 10) — стоимость билетов; 10 целых значений rng.integers(1000, 10000, 10) — номера рейсов. Создайте второй генератор с тем же seed=42 и убедитесь, что все три последовательности совпадают. Создайте третий генератор без seed и покажите, что значения отличаются. В комментариях объясните: зачем нужен seed и как он обеспечивает воспроизводимость экспериментов; чем default_rng отличается от старого np.random.seed (какой BitGenerator используется по умолчанию); почему в новом коде рекомендуется Generator, а не глобальный np.random; как SeedSequence помогает организовать независимые потоки случайных чисел для параллельных вычислений; в каких ситуациях отсутствие фиксированного seed — это хорошо, а в каких — плохо.

```python

import numpy as np

rng = np.random.default_rng(seed=42)

probabilities = rng.random(10)
prices = rng.normal(15000, 5000, 10)
flights = rng.integers(1000, 10000, 10)

print("Вероятности:", probabilities)
print("Стоимость билетов:", prices)
print("Номера рейсов:", flights)

rng2 = np.random.default_rng(seed=42)

probabilities2 = rng2.random(10)
prices2 = rng2.normal(15000, 5000, 10)
flights2 = rng2.integers(1000, 10000, 10)

print("Совпадают вероятности:", np.array_equal(probabilities, probabilities2))
print("Совпадают цены:", np.array_equal(prices, prices2))
print("Совпадают рейсы:", np.array_equal(flights, flights2))

rng3 = np.random.default_rng()

probabilities3 = rng3.random(10)
prices3 = rng3.normal(15000, 5000, 10)
flights3 = rng3.integers(1000, 10000, 10)

print("Без seed совпадают вероятности:", np.array_equal(probabilities, probabilities3))
print("Без seed совпадают цены:", np.array_equal(prices, prices3))
print("Без seed совпадают рейсы:", np.array_equal(flights, flights3))

# 1. seed задаёт начальное состояние генератора случайных чисел.
#    При одинаковом seed и одинаковом порядке вызовов
#    получаются одинаковые последовательности.
# 2. default_rng создаёт отдельный генератор Generator,
#    который по умолчанию использует BitGenerator PCG64.
#    Старый np.random.seed управляет глобальным генератором
#    на основе MT19937.
# 3. Generator рекомендуется в новом коде, потому что
#    позволяет создавать независимые объекты генераторов
#    без изменения общего глобального состояния.
# 4. SeedSequence позволяет создавать дочерние последовательности
#    через spawn() для независимых потоков случайных чисел.
# 5. Фиксированный seed полезен для тестов и повторения экспериментов.
#    Без seed удобно получать новые случайные данные при каждом запуске.
```

```python

import numpy as np

days1 = np.arange(1, 31)
days2 = np.linspace(1, 30, 30)

print("arange:", days1)
print("linspace:", days2)

print("Длина arange:", len(days1))
print("Длина linspace:", len(days2))

print("Первый элемент:", days1[0], days2[0])
print("Последний элемент:", days1[-1], days2[-1])

check = np.arange(0, 1, 0.1)
print("Количество точек с дробным шагом:", len(check))

# 1. linspace надёжнее, когда нужно точное количество точек,
#    потому что количество элементов задаётся напрямую.
# 2. arange удобнее, когда важен конкретный шаг между значениями.
# 3. При дробном шаге arange возможны погрешности вычислений.
#    В некоторых случаях это приводит к неожиданной длине массива
#    и может вызвать сдвиг данных при дальнейших расчётах.
# 4. Длину массива проверяют через len().
#    Если длина не совпадает с ожидаемой, лучше использовать linspace
#    с заранее заданным количеством элементов.
```
