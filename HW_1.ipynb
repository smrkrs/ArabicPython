1)Задача 3

Составить программу перевода радианной меры угла в градусы, минуты и секунды.


```python
import math

radians = float(input("Введите угол в радианах: "))

degrees_full = radians * 180 / math.pi

degrees = int(degrees_full)

minutes_full = (degrees_full - degrees) * 60
minutes = int(minutes_full)

seconds = (minutes_full - minutes) * 60

print("Градусы:", degrees)
print("Минуты:", minutes)
print("Секунды:", round(seconds, 2))
```

2)Задача 5. Дата некоторого дня характеризуется двумя натуральными числами: m (порядковый номер месяца) и n (число). По заданным n и m определить:
а) дату предыдущего дня (принять, что n и m не характеризуют 1 января);
б) дату следующего дня (принять, что n и m не характеризуют 31 декабря)


```python
m = int(input("Введите номер месяца: "))
n = int(input("Введите число: "))

days_in_month = [31, 28, 31, 30, 31, 30,
                 31, 31, 30, 31, 30, 31]

# Предыдущий день
if n > 1:
    prev_day = n - 1
    prev_month = m
else:
    prev_month = m - 1
    prev_day = days_in_month[prev_month - 1]

# Следующий день
if n < days_in_month[m - 1]:
    next_day = n + 1
    next_month = m
else:
    next_day = 1
    next_month = m + 1

print("Предыдущая дата:", prev_day, ".", prev_month)
print("Следующая дата:", next_day, ".", next_month)
```

3)Задача 3
Найти сумму ряда с точностью ε = 10^-3.
Общий член ряда:
a_n = n! / n^n

```python
import math

eps = 0.001
n = 1
S = 0

a = math.factorial(n) / n**n

while a >= eps:
    S += a
    n += 1
    a = math.factorial(n) / n**n

print("Сумма ряда =", S)
```

4)Задача 1.	Два простых числа называются «близнецами», если они отличаются друг от друга на 2 (например, 41 и 43). Напечатать все пары «близнецов» из отрезка [n, 2n], где n — заданное натуральное число больше 2.

```python
def is_prime(x):
    if x < 2:
        return False

    for i in range(2, x):
        if x % i == 0:
            return False

    return True


n = int(input("Введите n (> 2): "))

for x in range(n, 2 * n - 1):
    if is_prime(x) and is_prime(x + 2):
        print(x, x + 2)
```

4)Задача 2.	На плоскости заданы своими координатами n точек. Составить функцию, определяющую между какими из пар точек самое большое расстояние. Указание: координаты точек занести в массив.

```python
import math

def max_distance(x, y):
    max_d = 0
    p1 = 0
    p2 = 1

    for i in range(len(x)):
        for j in range(i + 1, len(x)):
            d = math.sqrt((x[j] - x[i])**2 + (y[j] - y[i])**2)

            if d > max_d:
                max_d = d
                p1 = i
                p2 = j

    return p1, p2, max_d


n = int(input("Введите количество точек: "))

x = []
y = []

for i in range(n):
    xi = float(input("Введите x: "))
    yi = float(input("Введите y: "))

    x.append(xi)
    y.append(yi)

p1, p2, d = max_distance(x, y)

print("Самое большое расстояние между точками", p1 + 1, "и", p2 + 1)
print("Расстояние =", d)
```

4)Задача 3.	Составить программу вычисления наибольшего общего делителя двух натуральных чисел, используя рекурсивную подпрограмму.

```python
def gcd(a, b):
    if b == 0:
        return a
    else:
        return gcd(b, a % b)


a = int(input("Введите первое число: "))
b = int(input("Введите второе число: "))

result = gcd(a, b)

print("НОД =", result)
```
