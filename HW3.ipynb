Задача 1 (создание класса, конструктор, метод). В аэропорту ведётся учёт пассажиров. Создайте класс Passenger с полями name (имя), flight_number (номер рейса) и seat (место). Реализуйте конструктор __init__, который принимает эти три значения и сохраняет их в объекте. Добавьте метод info(), который возвращает строку с информацией о пассажире в формате: Имя — рейс №..., место .... Создайте один экземпляр класса, вызовите метод и выведите результат на экран. Ожидаемый вывод, например: Иван Петров — рейс №SU1234, место 12A. В комментариях поясните, чем класс отличается от объекта и зачем нужен параметр self в методах.

```python
class Passenger:
    # Класс — шаблон; объект — конкретный экземпляр этого класса
    def __init__(self, name, flight_number, seat):
        # self — объект, для которого вызван метод
        # __init__ инициализирует уже созданный объект
        self.name = name
        self.flight_number = flight_number
        self.seat = seat

    def info(self):
        return f"{self.name} — рейс №{self.flight_number}, место {self.seat}"


passenger = Passenger("Иван Петров", "SU1234", "12A")
print(passenger.info())
```

Задача 2 (инкапсуляция). У каждого пассажира есть номер паспорта, который нельзя менять или читать напрямую. Создайте класс Passport, в котором номер паспорта хранится в приватном поле __passport_id. Реализуйте метод set_passport_id(value), предназначенный только для установки номера при регистрации, и метод get_passport_id() для чтения значения. Повторная попытка изменить номер после регистрации должна игнорироваться, а не вызывать ошибку. Продемонстрируйте работу класса: создайте паспорт, установите номер, попробуйте изменить его ещё раз и убедитесь, что значение осталось прежним. Дополнительно покажите, что снаружи нельзя напрямую прочитать passport.__passport_id — будет ошибка AttributeError, но через name mangling (passport._Passport__passport_id) доступ всё же существует. Объясните в комментариях, почему «игнорировать без ошибки» иногда лучше, чем выбрасывать исключение, и почему приватность в Python — это скорее договорённость, чем жёсткая защита.

```python
class Passport:
    def __init__(self):
        self.__passport_id = None
        self.__registered = False

    def set_passport_id(self, value):
        if self.__registered:
            # Безвредный повтор запроса не прерывает программу.
            # В критических операциях молчаливое игнорирование не всегда уместно.
            return
        self.__passport_id = value
        self.__registered = True

    def get_passport_id(self):
        return self.__passport_id


passport = Passport()
passport.set_passport_id("1234567890")
print("После регистрации:", passport.get_passport_id())

passport.set_passport_id("9999999999")
print("После попытки изменения:", passport.get_passport_id())

try:
    print(passport.__passport_id)
except AttributeError as error:
    print("Ожидаемая ошибка AttributeError:", error)

# Python изменяет имя поля, но не обеспечивает защиту секретных данных.
# Приватность здесь — соглашение, а не запрет на любой доступ.
print("Через name mangling:", passport._Passport__passport_id)
```

Задача 6 (статические методы). Аэропорту нужна утилита для проверки номеров рейсов, которая не зависит от конкретного объекта и может вызываться сама по себе. Создайте класс AirportUtils со статическим методом is_valid_flight(flight), который возвращает True, если строка начинается с букв «SU» и её длина равна 6, и False — в противном случае. Вызовите метод, не создавая объекта класса, — напрямую через имя класса: AirportUtils.is_valid_flight("SU1234"). Поясните, чем статический метод отличается от обычного и от классового (см. задачу 15) и когда его стоит использовать.

```python
class AirportUtils:
    @staticmethod
    def is_valid_flight(flight):
        # Статический метод не получает ни self, ни cls.
        # Он подходит для проверки, не зависящей от состояния объекта.
        return (
            flight.startswith("SU")
            and len(flight) == 6
        )

# Обычный метод получает self и работает с объектом.
# Классовый метод получает cls и работает с классом.
print(AirportUtils.is_valid_flight("SU1234"))  # True
print(AirportUtils.is_valid_flight("SU123"))   # False
print(AirportUtils.is_valid_flight("AA1234"))  # False
print(AirportUtils.is_valid_flight("SUABCD"))  # True
```

Унаследуйте от него классы Pilot и Passenger, каждый из которых обязан реализовать метод describe(), возвращающий краткое описание участника. Проверьте, что объект абстрактного класса AirportMember создать нельзя (будет ошибка TypeError), а объекты наследников — можно. Подсказка: используйте from abc import ABC, abstractmethod. В комментариях объясните, что произойдёт, если наследник не реализует абстрактный метод.

```python
from abc import ABC, abstractmethod


class AirportMember(ABC):
    @abstractmethod
    def describe(self):
        pass


class Pilot(AirportMember):
    def __init__(self, name):
        self.name = name

    def describe(self):
        return f"Пилот: {self.name}"


class Passenger(AirportMember):
    def __init__(self, name):
        self.name = name

    def describe(self):
        return f"Пассажир: {self.name}"


# Наследник без реализации describe() тоже останется абстрактным.
# Попытка создать его объект завершится TypeError.
try:
    member = AirportMember()
except TypeError as error:
    print("Ожидаемая ошибка TypeError:", error)

pilot = Pilot("Алексей Орлов")
passenger = Passenger("Иван Петров")
print(pilot.describe())
print(passenger.describe())
```

Задача 10 (имитация перегрузки). Сотруднику стойки регистрации нужен метод для оформления регистрации, который мог бы работать в разных ситуациях: иногда известен только пассажир, иногда пассажир и рейс, иногда ещё и место. Важно понимать: в Python нет классической перегрузки методов, как в Java или C++, потому что один метод — это один объект-функция. Поэтому «перегрузку» имитируют через значения по умолчанию и *args. Создайте класс CheckInDesk с методом check_in(passenger, flight=None, seat=None), который в зависимости от переданных аргументов выводит одно из трёх сообщений: «Зарегистрирован пассажир: », «Пассажир зарегистрирован на рейс: » или «Пассажир зарегистрирован на рейс , место ». Продемонстрируйте все три варианта вызова. Объясните в комментариях, почему в Python выбрали именно такой подход.

```python
class Passenger:
    def __init__(self, name):
        self.name = name


class CheckInDesk:
    def check_in(self, passenger, flight=None, seat=None):

        if flight is None:
            print(f"Зарегистрирован пассажир: {passenger.name}")
        elif seat is None:
            print(f"Пассажир {passenger.name} зарегистрирован на рейс: {flight}")
        else:
            print(
                f"Пассажир {passenger.name} зарегистрирован "
                f"на рейс {flight}, место {seat}"
            )

# Одно имя метода связано с одной функцией, а не с набором сигнатур.
# Значения по умолчанию позволяют обслуживать разные варианты вызова.
# Для переменного числа позиционных аргументов также применяют *args.
passenger = Passenger("Иван Петров")
desk = CheckInDesk()

desk.check_in(passenger)
desk.check_in(passenger, "SU1234")
desk.check_in(passenger, "SU1234", "12A")
```

Задача 11 (магические методы __str__ и __repr__). Пассажиров часто выводят на экран — в списках регистрации, посадочных талонах, отчётах. По умолчанию print(passenger) выводит нечитаемый адрес объекта в памяти, что неудобно. Добавьте в класс Passenger магический метод __str__, который возвращает строку вида «Пассажир "Иван Петров" (рейс SU1234, место 12A)». Убедитесь, что при вызове print(passenger) выводится именно эта строка. Дополнительно реализуйте метод __repr__, который возвращает техническое представление объекта, например Passenger('Иван Петров', 'SU1234', '12A') — оно используется при выводе списка пассажиров и в отладчике. Поясните разницу между __str__ и __repr__ и когда какой используется: __str__ — для человека, __repr__ — для разработчика.

```python
class Passenger:
    def __init__(self, name, flight_number, seat):
        self.name = name
        self.flight_number = flight_number
        self.seat = seat

    def __str__(self):
        # Представление для человека: используется print(passenger).
        return (
            f'Пассажир "{self.name}" '
            f'(рейс {self.flight_number}, место {self.seat})'
        )

    def __repr__(self):
        # Техническое представление: repr(passenger) и элементы списка.
        # !r добавляет представление строк с кавычками и экранированием.
        return (
            f"Passenger({self.name!r}, "
            f"{self.flight_number!r}, {self.seat!r})"
        )


passenger = Passenger("Иван Петров", "SU1234", "12A")
print(passenger)
print(repr(passenger))
```

Нужно, чтобы программа понимала, что это «один и тот же пассажир», и не дублировала его в списках. Реализуйте в классе Passenger магический метод __eq__, чтобы два пассажира считались равными, если совпадают их имя и номер рейса. Дополнительно реализуйте __hash__, чтобы пассажиров можно было хранить во множестве set и использовать как ключи в словаре. Проверьте, что два одинаковых пассажира попадают в множество как один элемент, а разные — как два. Объясните в комментариях, почему при переопределении __eq__ нужно также переопределять __hash__ — что произойдёт, если этого не сделать.

```python
class Passenger:
    def __init__(self, name, flight_number, seat):
        self.name = name
        self.flight_number = flight_number
        self.seat = seat

    def __eq__(self, other):
        return (
            self.name == other.name
            and self.flight_number == other.flight_number
        )

    def __hash__(self):
        return hash((self.name, self.flight_number))


p1 = Passenger("Иван Петров", "SU1234", "12A")
p2 = Passenger("Иван Петров", "SU1234", "15B")
p3 = Passenger("Анна Смирнова", "SU9999", "1A")

print(p1 == p2)
print(len({p1, p2}))
print(len({p1, p3}))

# Равные пассажиры должны иметь одинаковый хеш,
# поэтому используем те же поля, что и в __eq__.
# Если определить __eq__, но не определить __hash__,
# добавление объекта в set или использование ключом словаря
# вызовет TypeError.
```

Задача 13 (магические методы __add__ и __len__). В аэропорту есть списки пассажиров по рейсам, и иногда нужно объединять их (например, при объединении рейсов) или узнавать, сколько пассажиров в списке. Создайте класс Flight (рейс), в котором хранится список пассажиров. Реализуйте магический метод __add__ для объединения двух рейсов в один (возвращается новый рейс со всеми пассажирами) и метод __len__, возвращающий количество пассажиров в рейсе. Продемонстрируйте работу: создайте два рейса, объедините их через flight3 = flight1 + flight2 и выведите len(flight3). Объясните, почему __add__ возвращает новый объект, а не изменяет существующий, и как магические методы делают объекты «похожими» на встроенные типы.

```python
class Flight:
    def __init__(self, passengers):
        self.passengers = passengers

    def __add__(self, other):
        # Возвращаем новый рейс, чтобы не изменять исходные.
        return Flight(self.passengers + other.passengers)

    def __len__(self):
        return len(self.passengers)


# Благодаря специальным методам работают обычные операции + и len().
flight1 = Flight(["Иван Петров", "Анна Смирнова"])
flight2 = Flight(["Олег Соколов"])
flight3 = flight1 + flight2

print("Пассажиров в новом рейсе:", len(flight3))
print("Новый список:", flight3.passengers)
print("Первый рейс:", flight1.passengers)
print("Второй рейс:", flight2.passengers)
```

Объясните, чем @property удобнее классических геттеров и сеттеров, и что такое «вычисляемый атрибут».

```python
class Passenger:
    def __init__(self, name, flight_number, seat):
        self.name = name
        self.flight_number = flight_number
        self.seat = seat

    @property
    def seat(self):
        return self._seat

    @seat.setter
    def seat(self, value):
        if not isinstance(value, str) or value == "":
            raise ValueError("Место должно быть непустой строкой.")

        if not (
            value.isdigit()
            or (value[:-1].isdigit() and value[-1].isalpha())
        ):
            raise ValueError("Нужны цифры и одна буква в конце.")

        self._seat = value


#property удобнее обычных get_seat() и set_seat(),
# потому что позволяет работать с seat как с обычным атрибутом:
# passenger.seat и passenger.seat = "12A",
# но при этом внутри можно выполнять проверки.
#
# Вычисляемый атрибут — это свойство, значение которого не хранится отдельно,
# а вычисляется каждый раз при обращении к нему.

passenger = Passenger("Иван Петров", "SU1234", "7B")

for value in ["", 12, "12AB", "12A", "7B", "23"]:
    try:
        passenger.seat = value
        print("Место установлено:", passenger.seat)
    except ValueError as error:
        print("Ошибка:", error)
```

Задача 15 (классовый метод @classmethod). В аэропорт часто поступают данные о пассажирах в виде строк — например, из файла или базы данных: "Иван Петров;SU1234;12A". Добавьте в класс Passenger классовый метод from_string(cls, data), который создаёт объект пассажира из такой строки, разбивая её по точке с запятой. Продемонстрируйте создание пассажира через этот метод и вывод результата. Объясните разницу между @classmethod, @staticmethod и обычным методом: обычный метод получает первым аргументом экземпляр (self) и работает с конкретным объектом, @classmethod получает сам класс (cls) и часто используется как альтернативный конструктор, @staticmethod не получает ни self, ни cls и представляет собой просто утилиту. Расскажите, когда какой использовать.

```python
class Passenger:
    def __init__(self, name, flight_number, seat):
        self.name = name
        self.flight_number = flight_number
        self.seat = seat

    @classmethod
    def from_string(cls, data):
        name, flight_number, seat = data.split(";")
        return cls(name, flight_number, seat)

    def info(self):
        return f"{self.name} — рейс №{self.flight_number}, место {self.seat}"

# Обычный метод получает self и работает с конкретным объектом.
# classmethod получает cls: удобно создавать объект другим способом.
# staticmethod не получает ни self, ни cls: подходит для отдельных проверок.
passenger = Passenger.from_string("Иван Петров;SU1234;12A")
print(passenger.info())
```

```python
class FlightNotFoundError(Exception):
    pass

class Passenger:
    def __init__(self, name):
        self.name = name

class Airport:
    def __init__(self):
        self.flights = {"SU1234": [], "SU5678": []}

    def check_in_passenger(self, passenger, flight_number):
        if flight_number not in self.flights:
            raise FlightNotFoundError(f"Рейс {flight_number} не найден.")

        self.flights[flight_number].append(passenger)
        print(f"{passenger.name} зарегистрирован на рейс {flight_number}.")


# Своё исключение позволяет обработать именно отсутствие рейса,
# не перехватывая другие, не связанные с этим ошибки.
airport = Airport()
passenger = Passenger("Иван Петров")

airport.check_in_passenger(passenger, "SU1234")

try:
    airport.check_in_passenger(passenger, "SU9999")
except FlightNotFoundError as error:
    print("Ошибка:", error)

print("Пассажиров на SU1234:", len(airport.flights["SU1234"]))
```

При таком объёме важна экономия памяти: каждый объект Passenger по умолчанию хранит словарь атрибутов, что занимает много места. Добавьте в класс Passenger атрибут __slots__ = ("name", "flight_number", "seat"). Убедитесь, что теперь нельзя добавить новый атрибут объекту (например, passenger.email = "..." вызовет AttributeError). Сравните размер объекта с __slots__ и без него через sys.getsizeof — но учтите, что getsizeof показывает размер только «оболочки» объекта, не считая содержимое словаря; для честного сравнения используйте pympler.asizeof или оцените по количеству полей.
```python
import sys

class PassengerWithoutSlots:
    def __init__(self, name, flight_number, seat):
        self.name = name
        self.flight_number = flight_number
        self.seat = seat

class Passenger:
    __slots__ = ("name", "flight_number", "seat")

    def __init__(self, name, flight_number, seat):
        self.name = name
        self.flight_number = flight_number
        self.seat = seat


# __slots__ экономит память за счёт отсутствия __dict__ у этих объектов.
# Это полезно для большого количества объектов с одинаковыми полями.
# Наследование возможно, но наследник без своих __slots__ может получить __dict__.
# property тоже работает; свойство и слот не должны иметь одинаковое имя.
regular = PassengerWithoutSlots("Иван Петров", "SU1234", "12A")
passenger = Passenger("Иван Петров", "SU1234", "12A")

try:
    passenger.email = "passenger@example.com"
except AttributeError as error:
    print("Ошибка:", error)

# У обычного объекта дополнительно учитываем его словарь.
# Это оценка объекта и хранения атрибутов, без размеров самих строк.
regular_size = sys.getsizeof(regular) + sys.getsizeof(regular.__dict__)
slots_size = sys.getsizeof(passenger)

print("Без __slots__ (объект + словарь):", regular_size, "байт")
print("С __slots__:", slots_size, "байт")
```

Задача 22 (дескрипторы). Свойство @property — это удобно, но что если таких свойств с одинаковой проверкой нужно много? Лучше вынести логику в отдельный класс-дескриптор. Реализуйте собственный дескриптор RangeDescriptor(min_value, max_value), который хранит значение в instance.__dict__ (иначе все объекты делили бы одно значение) и проверяет, что оно в диапазоне, выбрасывая ValueError при нарушении. Используйте этот дескриптор в классе Passenger для полей age (например, диапазон 0–120) и baggage_weight (диапазон 0–32), продемонстрировав, что один и тот же дескриптор переиспользуется в разных полях. Объясните, как дескрипторы работают «под капотом» @property, зачем нужен метод __set_name__ и в чём преимущество дескриптора перед @property, если одинаковая проверка повторяется в десяти полях.

```python
class RangeDescriptor:
    def __init__(self, min_value, max_value):
        self.min_value = min_value
        self.max_value = max_value

    def __set_name__(self, owner, name):
        # Получаем имя поля: age или baggage_weight.
        self.name = name

    def __get__(self, instance, owner):
        if instance is None:
            return self
        return instance.__dict__[self.name]

    def __set__(self, instance, value):
        if not self.min_value <= value <= self.max_value:
            raise ValueError(
                f"{self.name}: значение должно быть "
                f"от {self.min_value} до {self.max_value}."
            )

        # Значение хранится у пассажира, а не в общем дескрипторе.
        instance.__dict__[self.name] = value


class Passenger:
    age = RangeDescriptor(0, 120)
    baggage_weight = RangeDescriptor(0, 32)

    def __init__(self, name, age, baggage_weight):
        self.name = name
        self.age = age
        self.baggage_weight = baggage_weight


# property тоже использует механизм дескрипторов.
# Здесь не нужно повторять одну проверку в отдельных сеттерах:
# для каждого поля создаётся свой экземпляр RangeDescriptor.
p1 = Passenger("Иван Петров", 25, 18)
p2 = Passenger("Анна Смирнова", 40, 12)

p1.age = 26

print("Первый пассажир:", p1.age, p1.baggage_weight)
print("Второй пассажир:", p2.age, p2.baggage_weight)

try:
    p1.age = 121
except ValueError as error:
    print("Ошибка:", error)

try:
    p1.baggage_weight = 33
except ValueError as error:
    print("Ошибка:", error)
```

Задача 23 (метаклассы). В аэропорту должен быть только один реестр рейсов — если создать второй, данные могут разойтись. Создайте метакласс SingletonMeta, который гарантирует, что у класса Airport будет только один экземпляр. Проверьте, что Airport() дважды возвращает один и тот же объект (сравните через is). Объясните, что такое метакласс, как он связан с классом (класс — это экземпляр метакласса) и зачем вообще нужны метаклассы, несмотря на их редкость в повседневной работе — в 95% проектов их не пишут, но понимать их важно для чтения чужого кода и фреймворков вроде Django.

```python
class SingletonMeta(type):
    _instances = {}

    def __call__(cls):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__()
        return cls._instances[cls]


class Airport(metaclass=SingletonMeta):
    pass


# Airport — экземпляр метакласса SingletonMeta.
# airport1 — экземпляр класса Airport.
# Метакласс позволяет задать общее поведение классов.
# Здесь он управляет созданием объектов и не допускает повторного создания.
# Это учебный пример для однопоточного запуска.
airport1 = Airport()
airport2 = Airport()

print(airport1 is airport2)
```

Задача 24 (паттерн Singleton). Реализуйте класс FlightRegistry через паттерн «Одиночка» двумя способами: через метакласс (как в задаче 23) и через переопределение метода __new__ (без метакласса). Сравните оба подхода: метакласс работает автоматически для наследников, но сложнее для понимания; __new__ проще читать, но приходится переопределять в каждом классе. Объясните, в каких случаях паттерн Singleton полезен (кэш, конфиг, логгер), а в каких — вреден (например, усложняет тестирование, скрывает зависимости и мешает подменять реализации).

```python
# Способ 1. Через метакласс.
class SingletonMeta(type):
    _instances = {}

    def __call__(cls):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__()
        return cls._instances[cls]


class FlightRegistry(metaclass=SingletonMeta):
    pass


registry1 = FlightRegistry()
registry2 = FlightRegistry()

print("Через метакласс:", registry1 is registry2)  # True


# Способ 2. Через __new__.
# Теперь определяем другую версию под тем же именем FlightRegistry.
class FlightRegistry:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance


registry3 = FlightRegistry()
registry4 = FlightRegistry()

print("Через __new__:", registry3 is registry4)  # True

# Метакласс удобно применять как общее правило, но он сложнее для понимания.
# __new__ позволяет описать Singleton непосредственно в одном классе.
# Singleton полезен, когда нужен один общий реестр, кеш или объект настроек.
# Минусы: общее состояние усложняет тесты, скрывает зависимости
# и мешает подменять объекты другими реализациями.
# Обе версии рассчитаны на однопоточный запуск.
```
