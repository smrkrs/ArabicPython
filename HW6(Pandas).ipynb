Задача 1 (создание Series, dtype, shape/size/ndim). Аэропорт ведёт учёт стоимости 7 авиабилетов: [25000, 18000, 12000, 9500, 31000, 15000, 22000] рублей. Создайте pd.Series из этих значений. Выведите dtype, itemsize, shape, nbytes. Дополнительно создайте второй Series из тех же значений с типом float32 через astype("float32") и сравните dtype, itemsize, nbytes с первым. Ожидаемый вывод: для первого — dtype=int64, itemsize=8, shape=(7,), nbytes=56; для второго — dtype=float32, itemsize=4, nbytes=28. В комментариях объясните: почему Pandas выбрал int64, а не int32; почему shape у одномерного Series — кортеж (7,), а не (7); что вернёт ticket_prices.iloc[7] и почему; как связаны itemsize, nbytes и size; во сколько раз float32 экономнее float64 и в каких случаях эта экономия критична (например, при хранении 10 млн проданных билетов за год); почему astype("float32") создаёт новый объект, а не меняет исходный Series на месте.

Задача 3 (подбор dtype, nullable-типы, переполнение). Аэропорт хранит три вида данных: возраст пассажира (0–110), вес багажа в килограммах (0–50), дальность рейса в километрах (100–15000). Создайте DataFrame из 100 строк. Понизьте типы: age → uint8, baggage_weight → uint8, distance → uint16 через astype(). Проверьте переполнение: np.uint8(200) + np.uint8(100). Создайте pd.Series([100, 200, None], dtype="Int64") и сравните с dtype="int64". Ожидаемый вывод: np.uint8(200) + np.uint8(100) = 44; Int64-Series сохраняет None как pd.NA, а int64-Series не может его создать; memory_usage(deep=True) после понижения типов сокращается примерно на 50%. В комментариях объясните: чем Int64 отличается от int64; зачем нужен pd.NA; как переполнение uint8 проявляется без ошибки (тихий wrap-around); почему для возраста пассажира и веса багажа uint8 подходит (максимум 255), а для дальности рейса нужен uint16 (максимум 65535); в каких случаях float32 недостаточно точен (дальность — целые километры, но при расчётах топлива нужна точность); как astype("category") дополнительно сокращает память для низкокардинальных строк (авиакомпании, направления, типы самолётов).

```python
import pandas as pd
import numpy as np

rng = np.random.default_rng(42)

df = pd.DataFrame({
    "age": rng.integers(0, 111, 100),
    "baggage_weight": rng.integers(0, 51, 100),
    "distance": rng.integers(100, 15001, 100)
})

print("Память до:", df.memory_usage(deep=True).sum())
print(df.dtypes)

df_small = df.astype({
    "age": "uint8",
    "baggage_weight": "uint8",
    "distance": "uint16"
})

print("Память после:", df_small.memory_usage(deep=True).sum())
print(df_small.dtypes)

print(np.uint8(200) + np.uint8(100))

s1 = pd.Series([100, 200, None], dtype="Int64")
print(s1)

try:
    s2 = pd.Series([100, 200, None], dtype="int64")
except (TypeError, ValueError) as e:
    print(type(e).__name__, e)

# Int64 допускает пропуски, int64 не поддерживает None при создании такого Series.
# pd.NA — специальное значение Pandas для отсутствующих данных.
# uint8 хранит числа от 0 до 255, поэтому подходит для возраста и багажа.
# uint16 хранит числа от 0 до 65535, поэтому подходит для дальности рейса.
# 200 + 100 = 300, но uint8 переполняется и возвращает 44 (может появиться предупреждение).
# Понижение типов сокращает использование памяти. Точный процент зависит от данных.
# float32 имеет ограниченную точность, что может быть недостаточно для расчётов топлива.
# astype("category") экономит память, когда в столбце мало уникальных строк.
```

Задача 4 (арифметика, выравнивание, assign, pd.col()). Создайте два Series выручки по рейсам с индексами ["SU100", "SU200", "SU300", "SU400"] и ["SU200", "SU300", "SU400", "SU500"]. Сложите их — зафиксируйте NaN на несовпадающих метках. Повторите через .add(other, fill_value=0). Создайте df = pd.DataFrame({"price": [25000, 18000, 12000], "passengers": [150, 200, 180]}). Через assign(revenue=lambda x: x["price"] * x["passengers"]) добавьте revenue. Через assign(revenue=pd.col("price") * pd.col("passengers")) — то же. Сравните результаты. Ожидаемый вывод: s1 + s2 даёт SU100=NaN, SU200=…, SU500=NaN; .add(..., fill_value=0) даёт числа везде; оба assign возвращают идентичные DataFrame с новым столбцом revenue. В комментариях объясните: почему выравнивание по меткам — ключевая особенность Pandas (данные из разных дней с разным набором рейсов); почему pd.col() помечен как «начальная поддержка»; почему assign не модифицирует исходный DataFrame (возвращает копию); как fill_value влияет на тип результата (может привести к float, если появляются NaN); в каких случаях lambda всё ещё нужен вместо pd.col() (сложные выражения, вызовы функций); что произойдёт при сложении Series с полностью непересекающимися индексами (все NaN).

```python
import pandas as pd

s1 = pd.Series(
    [1000, 2000, 3000, 4000],
    index=["SU100", "SU200", "SU300", "SU400"]
)

s2 = pd.Series(
    [500, 600, 700, 800],
    index=["SU200", "SU300", "SU400", "SU500"]
)

print(s1 + s2)
print(s1.add(s2, fill_value=0))

df = pd.DataFrame({
    "price": [25000, 18000, 12000],
    "passengers": [150, 200, 180]
})

df1 = df.assign(revenue=lambda x: x["price"] * x["passengers"])

df2 = df.assign(revenue=pd.col("price") * pd.col("passengers"))

print(df1)
print(df2)
print(df1.equals(df2))

# Pandas складывает Series по меткам индекса, а не по позициям.
# Это важно, если в разные дни набор рейсов отличается.
# Если метки нет в одном Series, при обычном сложении получается NaN.
# fill_value=0 заменяет отсутствующее значение нулём при вычислении.
# Результат может иметь тип float из-за объединения индексов и появления NaN.
# При полностью разных индексах обычное сложение даст только NaN.
# assign() возвращает новый DataFrame и не изменяет исходный.
# pd.col() — новый синтаксис Pandas 3.0, его поддержка пока развивается.
# lambda удобнее для сложных выражений и вызовов функций.
```

Задача 7 (память: memory_usage, Parquet, категории). Создайте DataFrame из 1 000 000 строк: flight_id (int64), delay_minutes (float64), passengers (int64). Выведите memory_usage(deep=True) в МБ. Понизьте типы через pd.to_numeric(..., downcast=...): flight_id → int32, delay_minutes → float32, passengers → uint16. Сравните память. Добавьте airline (Аэрофлот/S7/Победа/Уральские) из 4 значений — сравните memory_usage(deep=True) для str и category. Сохраните оба варианта в CSV и Parquet. Ожидаемый вывод: исходно ≈15.26 МБ, после понижения ≈4.77 МБ; category для airline экономит ≈96% памяти относительно str; Parquet в 2–5 раз меньше CSV и в 3–10 раз быстрее. В комментариях объясните: во сколько раз uint16 экономнее float64 (в 4 раза); в каких случаях category хуже str (высокая кардинальность — много уникальных номеров рейсов); почему memory_usage(deep=True) строкового столбца превышает nbytes (учитывает объекты Python-строк); как downcast выбирает тип (минимальный, вмещающий все значения); почему Parquet сохраняет category и datetime, а CSV — нет (Parquet хранит метаданные и типы).

```python
import pandas as pd
import numpy as np

rng = np.random.default_rng(42)
n = 1_000_000

df = pd.DataFrame({
    "flight_id": np.arange(1, n + 1, dtype="int64"),
    "delay_minutes": rng.uniform(0, 180, n),
    "passengers": rng.integers(1, 501, n, dtype="int64")
})

print("Исходная память, МБ:")
print(df.memory_usage(deep=True) / 1024**2)

df_small = df.copy()

df_small["flight_id"] = pd.to_numeric(df_small["flight_id"], downcast="integer")
df_small["delay_minutes"] = pd.to_numeric(df_small["delay_minutes"], downcast="float")
df_small["passengers"] = pd.to_numeric(df_small["passengers"], downcast="unsigned")

print("Память после оптимизации, МБ:")
print(df_small.memory_usage(deep=True) / 1024**2)
print(df_small.dtypes)

df_str = df_small.copy()
df_str["airline"] = np.resize(
    ["Аэрофлот", "S7", "Победа", "Уральские"], n
)

df_category = df_str.copy()
df_category["airline"] = df_category["airline"].astype("category")

print("Строки, МБ:", df_str["airline"].memory_usage(deep=True) / 1024**2)
print("Категории, МБ:", df_category["airline"].memory_usage(deep=True) / 1024**2)

df_str.to_csv("flights_str.csv", index=False)
df_category.to_csv("flights_category.csv", index=False)

df_str.to_parquet("flights_str.parquet", index=False)
df_category.to_parquet("flights_category.parquet", index=False)

# uint16 занимает 2 байта, float64 — 8 байт, экономия в 4 раза.
# downcast выбирает минимальный подходящий числовой тип.
# category экономит память при небольшом числе уникальных значений.
# При большом числе уникальных строк category может занимать больше памяти.
# deep=True учитывает память Python-строк, которую nbytes не учитывает полностью.
# Parquet хранит данные по столбцам, поддерживает сжатие и сохраняет типы.
# CSV не сохраняет информацию о category и datetime как о типах.
# Реальная экономия и размеры файлов зависят от данных и способа хранения.
# Для трёх исходных столбцов int64/float64/int64 память около 22.9 МиБ.
# После перехода на int32/float32/uint16 — около 9.54 МиБ, а не 4.77 МиБ.
```

```python
import pandas as pd
import numpy as np

rng = np.random.default_rng(42)

passengers = pd.DataFrame({
    "age": rng.integers(0, 91, 100),
    "city": rng.choice(["Москва", "Сочи", "Казань"], 100),
    "gender": rng.choice(["М", "Ж"], 100),
    "ticket_price": rng.integers(5000, 50001, 100)
})

passengers["age_group"] = pd.cut(
    passengers["age"],
    bins=[0, 18, 35, 60, 100],
    labels=["<18", "18-35", "35-60", ">60"],
    include_lowest=True
)

passengers["age_quantile"] = pd.qcut(
    passengers["age"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)

print(passengers.head())

print(passengers["age_group"].value_counts())
print(passengers["age_quantile"].value_counts())

print(pd.crosstab(
    index=passengers["city"],
    columns=passengers["age_group"],
    margins=True,
    normalize="index"
))

print(pd.pivot_table(
    passengers,
    values="ticket_price",
    index="city",
    columns="gender",
    aggfunc="mean",
    margins=True,
    fill_value=0
))

# cut() делит значения по заданным границам, группы могут быть разного размера.
# qcut() делит данные на группы примерно одинаковой численности.
# Если границы qcut совпадают, может возникнуть ValueError.
# duplicates="drop" позволяет удалить повторяющиеся границы.
# normalize="index" показывает доли внутри строк.
# normalize="columns" показывает доли внутри столбцов.
# normalize=True показывает доли от общей суммы.
# margins=True добавляет строку и при необходимости столбец All.
# pivot_table() умеет объединять повторяющиеся значения через aggfunc.
# Обычный pivot() не умеет агрегировать дубликаты.
# fill_value=0 заменяет пропуски после агрегации.
# При right=False интервалы cut не включают правую границу.
```

Задача 26 (groupby: transform/filter/apply). Используйте DataFrame из задачи 25. Примените .agg({"ticket_price": ["sum", "mean"], "passengers": ["count", "max"]}). Примените transform("mean") и transform(lambda x: x.fillna(x.mean())). Примените filter(lambda g: len(g) > 50). Примените apply(weighted_avg) с np.average(ticket_price, weights=passengers). Примените apply(..., include_groups=False). Ожидаемый вывод: transform("mean") возвращает Series той же длины; filter оставляет только группы с >50 строками; apply с include_groups=False не передаёт группировочные столбцы в функцию. В комментариях объясните: в каких случаях transform незаменим (когда нужен результат той же длины, например, заполнение пропусков групповым средним); почему apply медленнее встроенных методов; что делает include_groups=False (убирает группировочные столбцы из DataFrame внутри функции); как filter отличается от булевой маски (фильтрует целые группы, а не строки); когда agg со словарём предпочтительнее списка (разные функции для разных столбцов); почему transform возвращает Series той же длины, а agg — свёрнутый результат.

```python
import pandas as pd
import numpy as np

print(flights.groupby("airline").agg({
    "ticket_price": ["sum", "mean"],
    "passengers": ["count", "max"]
}))

print(flights.groupby("airline")["ticket_price"].transform("mean"))

print(flights.groupby("airline")["ticket_price"].transform(
    lambda x: x.fillna(x.mean())
))

print(flights.groupby("airline").filter(
    lambda g: len(g) > 50
))

def weighted_avg(group):
    valid = group["ticket_price"].notna()
    return np.average(
        group.loc[valid, "ticket_price"],
        weights=group.loc[valid, "passengers"]
    )

print(flights.groupby("airline").apply(weighted_avg))

print(flights.groupby("airline").apply(
    weighted_avg, include_groups=False
))

# agg() собирает статистики по каждой группе.
# transform() возвращает результат той же длины, что исходный столбец.
# transform("mean") повторяет среднее группы для каждой её строки.
# transform() удобен для заполнения пропусков средним по группе.
# filter() отбирает целые группы, а не отдельные строки.
# Здесь filter() оставляет только Aeroflot, так как в группе 60 строк.
# apply() позволяет использовать собственную функцию.
# apply() обычно медленнее встроенных методов Pandas.
# include_groups=False исключает столбец группировки из передаваемой группы.
# В Pandas 3.0 этот режим используется по умолчанию.
# agg() уменьшает количество строк, а transform() сохраняет их число.
# Словарь в agg() удобен, когда для разных столбцов нужны разные функции.
```

```python
import pandas as pd
import numpy as np

rng = np.random.default_rng(42)

passengers = pd.DataFrame({
    "age": rng.integers(0, 91, 100),
    "city": rng.choice(["Москва", "Сочи", "Казань"], 100),
    "gender": rng.choice(["М", "Ж"], 100),
    "ticket_price": rng.integers(5000, 50001, 100)
})

passengers["age_group"] = pd.cut(
    passengers["age"],
    bins=[0, 18, 35, 60, 100],
    labels=["<18", "18-35", "35-60", ">60"],
    include_lowest=True
)

passengers["age_quantile"] = pd.qcut(
    passengers["age"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)

print(passengers.head())

print(passengers["age_group"].value_counts())
print(passengers["age_quantile"].value_counts())

print(pd.crosstab(
    index=passengers["city"],
    columns=passengers["age_group"],
    margins=True,
    normalize="index"
))

print(pd.crosstab(
    index=passengers["city"],
    columns=passengers["age_group"],
    values=passengers["ticket_price"],
    aggfunc="sum"
))

print(pd.pivot_table(
    passengers,
    values="ticket_price",
    index="city",
    columns="gender",
    aggfunc=["sum", "mean"],
    margins=True,
    fill_value=0
))

print(pd.cut(
    passengers["age"],
    bins=[0, 18, 35, 60, 100],
    right=False
))

# cut() делит значения на интервалы по заданным границам.
# qcut() создаёт группы примерно одинаковой численности.
# При малом числе уникальных значений qcut() может вызвать ValueError.
# crosstab() строит таблицу частот.
# values и aggfunc позволяют вычислять сумму или другую статистику.
# normalize="index" показывает доли по строкам.
# normalize="columns" показывает доли по столбцам.
# margins=True добавляет итоговые значения All.
# pivot_table() объединяет данные с помощью агрегирующих функций.
# aggfunc=["sum", "mean"] создаёт MultiIndex столбцов.
# fill_value=0 заменяет пропуски в итоговой таблице.
# right=False создаёт интервалы [a, b), включая левую границу.
```

Ожидаемый вывод: rolling(7).mean() — первые 6 значений NaN; min_periods=1 — без NaN; center=True — метка в центре окна; rolling("30D") требует DatetimeIndex. В комментариях объясните: чем center=True отличается (окно центрируется на текущей точке, а не заканчивается в ней); когда raw=True ускоряет apply (передаёт NumPy-массив вместо Series); как win_type меняет веса (треугольное, гауссово окно); почему временные окна требуют DatetimeIndex; что произойдёт при min_periods > window (все значения NaN); как closed влияет на включение границ ("right", "left", "both", "neither").

```python
import pandas as pd
import numpy as np

rng = np.random.default_rng(42)

dates = pd.date_range("2025-01-01", periods=100, freq="D")

flights = pd.Series(
    rng.integers(50, 201, 100),
    index=dates
)

print(flights.rolling(7).mean())
print(flights.rolling(7, min_periods=1).mean())
print(flights.rolling(7, center=True).mean())

print(flights.rolling(7, win_type="triang").mean())

print(flights.rolling("30D").sum())

try:
    print(flights.rolling("5B").mean())
except ValueError:
    print("ValueError: rolling не поддерживает окно 5B")

print(flights.rolling(7).agg(["mean", "std"]))

print(flights.rolling(7).apply(
    lambda x: np.percentile(x, 90),
    raw=True
))

# rolling(7) использует окно из 7 наблюдений.
# Первые 6 значений будут NaN, так как окно ещё не заполнено.
# min_periods=1 позволяет считать среднее с первого элемента.
# center=True центрирует окно относительно текущей точки.
# win_type="triang" задаёт треугольные веса.
# rolling("30D") использует временное окно в 30 дней.
# Временные окна требуют индекса с датами.
# Частота "5B" не поддерживается как rolling-окно.
# raw=True передаёт в apply() NumPy-массив вместо Series.
# Если min_periods больше размера числового окна, возникает ValueError.
# closed определяет, какие границы окна включаются.
# Возможные значения: right, left, both, neither.
```

В комментариях объясните: чем expanding отличается от rolling (накопительное окно от начала vs скользящее фиксированное); как adjust=False меняет формулу (рекурсивное взвешивание без деления на сумму весов); почему transform нужен для groupby().rolling() (чтобы вернуть исходную форму); как halflife связан с span (halflife = ln(2)/α, span = 2/α - 1); что произойдёт при ewm без параметров (span=1 по умолчанию); как ignore_na=True влияет на результат (пропуски не влияют на веса).

```python
import pandas as pd
import numpy as np

rng = np.random.default_rng(42)

s = pd.Series(rng.integers(50, 201, 30))

print(s.expanding().mean())
print(s.expanding().sum())

print(s.ewm(span=10).mean())
print(s.ewm(halflife=5).mean())
print(s.ewm(alpha=0.3).mean())
print(s.ewm(span=10, adjust=False).mean())

df = pd.DataFrame({
    "airline": ["Aeroflot"] * 15 + ["S7"] * 15,
    "passengers": rng.integers(50, 201, 30)
})

df["rolling_mean"] = df.groupby("airline")["passengers"].transform(
    lambda x: x.rolling(7).mean()
)

print(df)

s_missing = pd.Series([10, np.nan, 20])

print(s_missing.ewm(span=3, ignore_na=True).mean())

# expanding() использует все значения с начала ряда до текущего.
# rolling() использует скользящее окно фиксированного размера.
# ewm() вычисляет экспоненциально взвешенное среднее.
# span=10 соответствует alpha = 2 / 11, примерно 0.18.
# Чем больше alpha, тем сильнее влияние новых значений.
# halflife задаёт скорость уменьшения веса старых наблюдений.
# Связь параметров: alpha = 1 - 0.5 ** (1 / halflife).
# span = 2 / alpha - 1.
# adjust=False использует рекурсивное вычисление среднего.
# transform() сохраняет количество строк и исходные индексы.
# Без задания span, alpha, halflife или com ewm() выдаст ValueError.
# ignore_na=True не учитывает пропуски при расчёте относительных весов.
```

```python
import pandas as pd

df1 = pd.DataFrame({
    "flight_id": ["SU101", "SU102"],
    "price": [15000, 20000]
})

df2 = pd.DataFrame({
    "flight_id": ["SU103", "SU104"],
    "price": [18000, 25000]
})

print(pd.concat([df1, df2]))

print(pd.concat([df1, df2], ignore_index=True))

print(pd.concat([df1, df2], axis=1))

print(pd.concat([df1, df2], keys=["a", "b"]))

try:
    print(pd.concat(
        [df1, df2],
        ignore_index=True,
        keys=["a", "b"]
    ))
except ValueError:
    print("ValueError: нельзя одновременно использовать ignore_index=True и keys")

df = pd.DataFrame({
    "destination": ["Москва", "Сочи", "Дубай"],
    "Q1": [150, 200, 180],
    "Q2": [170, 220, 190]
})

print(df)

print(df.stack())
print(df.unstack())

print(pd.melt(
    df,
    id_vars=["destination"],
    value_vars=["Q1", "Q2"],
    var_name="quarter",
    value_name="passengers"
))

# concat() объединяет несколько DataFrame.
# Без ignore_index сохраняются старые индексы, даже если они повторяются.
# ignore_index=True создаёт новый последовательный индекс.
# axis=1 объединяет таблицы по столбцам.
# keys создаёт дополнительный уровень индекса MultiIndex.
# ignore_index=True и keys нельзя использовать одновременно.
# stack() переносит столбцы в индекс.
# stack(dropna=False) предназначен для сохранения пропусков в старом режиме stack.
# unstack() переносит уровень индекса в столбцы.
# Несуществующий уровень unstack() вызывает ошибку.
# melt() превращает широкую таблицу в длинную.
# melt() удобен для преобразования нескольких столбцов value_vars.
```

Задача 32 (merge/join/anti-join). Создайте flights и passengers. Примените merge(how="inner"), "left", "right", "outer", "left_anti", "right_anti". Используйте on="flight_id", left_on/right_on, suffixes=("_f", "_p"), indicator=True. Проверьте validate="one_to_one" — зафиксируйте MergeError при дубликатах. Создайте flights.set_index("flight_id").join(passengers.set_index("flight_id")). Ожидаемый вывод: inner оставляет только совпадающие ключи; left_anti возвращает рейсы без пассажиров; indicator=True добавляет столбец _merge; validate="one_to_one" вызывает MergeError при дубликатах. В комментариях объясните: чем merge отличается от join (join — по индексу, merge — по столбцам); когда anti-join полезен (поиск «осиротевших» рейсов без пассажиров); что делает indicator=True (добавляет _merge с "both"/"left_only"/"right_only"); как suffixes предотвращает конфликт имён; почему validate защищает от декартова произведения; как merge обрабатывает NaN в ключах (не совпадают, остаются в left_only/right_only).

```python
import pandas as pd
from pandas.errors import MergeError

flights = pd.DataFrame({
    "flight_id": ["SU101", "SU102", "SU103", "SU104"],
    "airline": ["Aeroflot", "S7", "Pobeda", "S7"],
    "status": ["вылет", "вылет", "задержан", "вылет"]
})

passengers = pd.DataFrame({
    "flight_id": ["SU101", "SU101", "SU103", "SU105"],
    "passenger_id": [1, 2, 3, 4],
    "status": ["регистрация", "регистрация", "посадка", "регистрация"]
})

print(flights.merge(
    passengers, how="inner", on="flight_id",
    suffixes=("_f", "_p")
))

print(flights.merge(
    passengers, how="left", on="flight_id",
    suffixes=("_f", "_p")
))

print(flights.merge(
    passengers, how="right", on="flight_id",
    suffixes=("_f", "_p")
))

print(flights.merge(
    passengers, how="outer", on="flight_id",
    suffixes=("_f", "_p"), indicator=True
))

print(flights.merge(
    passengers, how="left_anti", on="flight_id"
))

print(flights.merge(
    passengers, how="right_anti", on="flight_id"
))

passengers2 = passengers.rename(columns={"flight_id": "flight_code"})

print(flights.merge(
    passengers2,
    left_on="flight_id",
    right_on="flight_code",
    suffixes=("_f", "_p")
))

try:
    print(flights.merge(
        passengers,
        on="flight_id",
        validate="one_to_one"
    ))
except MergeError:
    print("MergeError: найдены дубликаты ключей")

print(flights.set_index("flight_id").join(
    passengers.set_index("flight_id"),
    lsuffix="_f",
    rsuffix="_p"
))

# inner оставляет только совпадающие ключи.
# left сохраняет все ключи левой таблицы.
# right сохраняет все ключи правой таблицы.
# outer объединяет все ключи двух таблиц.
# left_anti возвращает рейсы без записей о пассажирах.
# right_anti возвращает пассажиров без соответствующего рейса.
# merge() обычно соединяет данные по столбцам, join() — по индексам.
# left_on и right_on позволяют соединять столбцы с разными названиями.
# suffixes предотвращает конфликты одинаковых имён столбцов.
# indicator=True добавляет _merge со значениями both, left_only, right_only.
# validate="one_to_one" проверяет уникальность ключей.
# При дубликатах validate защищает от случайного размножения строк.
# В Pandas при merge пропущенные ключи NaN могут совпадать друг с другом.
```
